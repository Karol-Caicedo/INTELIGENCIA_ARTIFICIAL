#**¿QUÉ ES PANDAS?**


Pandas es la biblioteca más popular y fundamental para la Manipulación y el Análisis. Es una biblioteca para Ciencia de Datos de código abierto (open source), construida sobre el lenguaje Python, y que proporciona un enfoque rápido y flexible, con estructuras robustas para trabajar con datos relacionales (o etiquetados), y todo eso de manera sencilla y intuitiva.

---




#**¿PARA QUÉ SIRVE?**

Pandas es extremadamente versátil y se utiliza para:

*   **Limpieza y tratamiento de datos:** Manejar valores nulos o faltantes, eliminar duplicados y transformar formatos.
*   **Análisis Exploratorio de Datos (EDA):** Obtener estadísticas descriptivas rápidas (medias, distribuciones) para entender los datos.
*   **Soporte en Machine Learning:** Preparar y moldear los conjuntos de datos (*feature engineering*) antes de entrenar modelos.
*   **Consultas y *queries* (Estilo SQL):** Filtrar, agrupar (`groupby`) y combinar (`merge`/`join`) tablas de forma eficiente.
*   **Visualización de datos:** Generar gráficos rápidos (líneas, barras, histogramas) directamente desde los datos.
*   **Web Scraping:** Extraer tablas de datos estructuradas directamente desde páginas web.

#**OBEJTIVO PRINCIPAL DE LA LIBRERÍA PANDAS**

El objetivo principal de Pandas es ofrecer estructuras de datos de alto rendimiento, eficientes y fáciles de usar, diseñadas específicamente para limpiar, manipular, transformar y analizar datos estructurados (tabulares, relacionales o en series temporales).

Busca simplificar todo el proceso de preparación de datos, permitiendo pasar de datos en bruto (*raw data*) a datos listos para el análisis o el modelado en solo unas pocas líneas de código.

# **PRINCIPALES FUNCIONES Y CARACTERÍSTICAS**

* **Estructuras de Datos Robustas:** Introduce el *DataFrame* (tablas de dos dimensiones con filas y columnas) y la *Series* (columnas unidimensionales), capaces de manejar diversos tipos de datos simultáneamente.

* **Alineación de Datos Inteligente:** Alinea de forma automática los datos en cálculos y operaciones, gestionando de manera nativa las etiquetas de las filas y columnas.

* **Tratamiento de Datos Faltantes:** Ofrece herramientas integradas y sencillas para detectar, filtrar y rellenar valores nulos o datos perdidos (`NaN`).

* **Indexación y Segmentación Avanzada:** Permite seleccionar, dividir (*slicing*) y trabajar subconjuntos de grandes volúmenes de datos basados en etiquetas o posiciones.

* **Operaciones de Agrupación y Agregación:** Implementa la funcionalidad `groupby` para dividir, aplicar funciones estadísticas y combinar conjuntos de datos de forma rápida, similar a SQL.

* **Fusión y Unión de Tablas:** Facilita la combinación de múltiples fuentes de datos mediante operaciones de unión (`merge` y `join`) de alta velocidad.

* **Soporte de Series Temporales:** Incluye herramientas avanzadas para trabajar con fechas, periodos de tiempo y frecuencias de datos como remuestreo y ventanas móviles.

*   **Código abierto (Open Source):** Construida sobre el lenguaje Python.

*   **Enfoque rápido y flexible:** Diseñada para el alto rendimiento.

*   **Estructuras robustas:** Ideal para trabajar con datos relacionales o etiquetados.

*   **Uso intuitivo:** Permite realizar procesos complejos de manera sencilla.

# **Funciones comunes de Pandas**

| Función | Descripción |
|---|---|
| `read_csv()` | Permite leer archivos CSV y cargarlos en un DataFrame. |
| `head()` | Muestra las primeras 5 filas del DataFrame. |
| `info()` | Proporciona información general sobre los datos y tipos de columnas. |
| `describe()` | Genera estadísticas descriptivas básicas de los datos numéricos. |
| `groupby()` | Agrupa datos para realizar operaciones estadísticas o de análisis. |
| `sort_values()` | Ordena los datos según una o varias columnas. De menor a mayor → `ascending=True` y de mayor a menor → `ascending=False`. |
| `dropna()` | Elimina filas o columnas con valores nulos o vacíos. |
| `fillna()` | Rellena valores faltantes con datos específicos. |
| `merge()` | Combina dos DataFrames utilizando columnas en común. |
| `loc[]` y `iloc[]` | Permiten seleccionar datos por etiquetas o posiciones. |

##**IMPORTACION DE LA LIBRERIA PANDAS Y EL DATASET**

```python
import pandas as pd

url = "https://raw.githubusercontent.com/Karol-Caicedo/INTELIGENCIA_ARTIFICIAL/main/SEGUNDO_CORTE/TALLERES/ventas_tienda.csv"
```

```python
df = pd.read_csv(url, index_col=0)
df
```

## **OPERACIÓN ENTRE COLUMNAS**

```python
#CREAR LA COLUMNA TOTAL DE VENTA
df["Total_Venta"] = (df["Cantidad"] * df["Precio_Unitario"]) - df["Descuento"]
df
```

```python
#TOTAL VENDIDO POR CIUDAD ORGANIZADO DE MAYOR A MENOR
ventas_ciudad = df.groupby("Ciudad")["Total_Venta"].sum().sort_values(ascending=False)
ventas_ciudad
```

```python
#TOTAL VENDIDO POR CIUDAD ORGANIZADO DE MENOR A MAYOR
ventas_ciudad = df.groupby("Ciudad")["Total_Venta"].sum().sort_values(ascending=True)
ventas_ciudad
```

```python
#PRECIO PROMEDIO POR CATEGORIAS DE MAYOR A MENOR
promedio_categoria = df.groupby("Categoria")["Precio_Unitario"].mean().sort_values(ascending=False)
promedio_categoria
```

```python
#MOSTRAR PROMEDIO POR CATEGORIAS MENOR A MAYOR
promedio_categoria = df.groupby("Categoria")["Precio_Unitario"].mean().sort_values()
promedio_categoria
```

```python
#CANTIDAD TOTAL DE PRODUCTOS VENDIDOS
productos_vendidos = df.groupby("Producto")["Cantidad"].sum()
productos_vendidos
```

```python
metodos_pago = df["Metodo_Pago"].value_counts()
metodos_pago
```

##**VISUALIZACION DE LAS OPERACIONES ENTRE COLUMNAS CON MATPLOTLIB**

```python
import matplotlib.pyplot as plt
```

```python
ventas_ciudad = df.groupby("Ciudad")["Total_Venta"].sum().sort_values(ascending=True)

ventas_ciudad.plot(kind="bar")

plt.title("Total Vendido por Ciudad")
plt.xlabel("Ciudad")
plt.ylabel("Total de Ventas")

plt.xticks(rotation=0)

plt.show()
```

```python
productos_vendidos = df.groupby("Producto")["Cantidad"].sum()

productos_vendidos.sort_values(ascending=False).plot(kind="bar")

plt.title("Productos Más Vendidos")
plt.xlabel("Producto")
plt.ylabel("Cantidad Vendida")

plt.show()
```

```python
promedio_categoria = df.groupby("Categoria")["Precio_Unitario"].mean().sort_values(ascending=False)

promedio_categoria.plot(kind="bar")

plt.title("Promedio de Precios por Categoría")
plt.xlabel("Categoría")
plt.ylabel("Precio Promedio")

plt.xticks(rotation=0)
plt.show()
```

```python
metodos_pago = df["Metodo_Pago"].value_counts()

metodos_pago.plot(kind="pie", autopct="%1.1f%%")

plt.title("Métodos de Pago")
plt.ylabel("")

plt.show()
```
