## **1. Investigación Conceptual**

## **NumPy**
### **1. ¿Qué es NumPy?**
Es la biblioteca fundamental para la computación científica en Python. Proporciona soporte para crear y manipular arreglos y matrices multidimensionales de gran tamaño, junto con una amplia colección de funciones matemáticas de alto nivel para operar con estas estructuras de manera eficiente

### **2. ¿Por qué NumPy es más eficiente que las listas tradicionales de Python?**
Las listas de Python almacenan punteros a objetos dispersos en la memoria, lo que requiere validación de tipos en cada operación. NumPy utiliza arreglos homogéneos (datos del mismo tipo) almacenados en bloques de memoria contiguos, lo que permite aprovechar el procesamiento vectorizado y el caché de la CPU. Además, el núcleo de NumPy está escrito en lenguajes de bajo nivel como C y Fortran, acelerando drásticamente los cálculos.

### **3. ¿Qué es un ndarray?**
Es el objeto principal de NumPy. Se trata de un arreglo multidimensional (N-dimensional array) de elementos del mismo tipo y tamaño fijo en memoria. Funciona como un contenedor rápido y flexible para grandes conjuntos de datos.  

### 4. **¿Qué relación existe entre NumPy y el álgebra lineal?**
NumPy está profundamente ligado al álgebra lineal, ya que incluye el módulo numpy.linalg, el cual permite realizar operaciones complejas como multiplicaciones matriciales, cálculo de determinantes, inversiones de matrices y descomposiciones. Estas operaciones son la base matemática de los algoritmos de aprendizaje automático.  

### **5. ¿Cómo utilizan los modelos de Deep Learning las matrices de NumPy?**
En Deep Learning, los datos (como imágenes, texto o sonido), los pesos y los sesgos de las redes neuronales se representan como tensores (matrices multidimensionales). Las matrices de NumPy actúan como la estructura subyacente inicial para almacenar estos datos y realizar las operaciones de propagación hacia adelante y hacia atrás (backpropagation) de manera vectorizada antes de pasarlos a frameworks especializados como TensorFlow o PyTorch.

```python
import numpy as np

# Creación de la matriz inicial según la guía
matriz = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print("Matriz original:")
print(matriz)
print("-" * 20)

# 1. Mostrar el elemento central
# El elemento central está en la fila 1, columna 1 (recordando que el índice empieza en 0)
elemento_central = matriz[1, 1]
print(f"Elemento central: {elemento_central}")

# 2. Mostrar la segunda fila
# Seleccionamos la fila en el índice 1 y todas las columnas (:)
segunda_fila = matriz[1, :]
print(f"Segunda fila: {segunda_fila}")

# 3. Mostrar la tercera columna
# Seleccionamos todas las filas (:) y la columna en el índice 2
tercera_columna = matriz[:, 2]
print(f"Tercera columna: {tercera_columna}")
```

## **2. Operaciones Matemáticas**

En el análisis de datos y la Inteligencia Artificial, el primer paso para comprender un conjunto de datos es calcular sus estadísticos descriptivos básicos. NumPy nos proporciona funciones altamente optimizadas para realizar estos cálculos de forma inmediata sobre sus arreglos (ndarrays).

*np.mean()* **(Promedio/Media)**: Calcula el valor central de los datos, sumando todos los elementos y dividiéndolos por la cantidad total.

*np.max()* y *np.min()* **(Máximo y Mínimo)**: Identifican los valores extremos del conjunto de datos, lo que nos ayuda a conocer el rango completo de nuestra información.

*np.std()* **(Desviación Estándar)**: Mide qué tan dispersos están los datos con respecto al promedio.

```python
import numpy as np

# Definición del arreglo de datos según el taller
datos = np.array([12, 18, 20, 25, 30, 35, 40])

# Cálculo e impresión de los estadísticos descriptivos
print("--- Estadísticos Descriptivos ---")
print(f"Promedio (Media): {np.mean(datos):.2f}")
print(f"Valor Máximo:     {np.max(datos)}")
print(f"Valor Mínimo:     {np.min(datos)}")
print(f"Desviación Est.:  {np.std(datos):.2f}")
```

## **Respuestas a las Preguntas de Análisis**

### ***¿Qué representa la desviación estándar?***
Representa la medida de dispersión o variabilidad de un conjunto de datos en relación con su media (promedio). Matemáticamente, indica en promedio cuánto se alejan los valores individuales del valor central.  
* Una desviación estándar baja significa que la mayoría de los datos están agrupados muy cerca del promedio (alta consistencia).
* Una desviación estándar alta indica que los datos están extendidos sobre un rango de valores más amplio (alta variabilidad).

### ***¿Por qué es importante conocer la dispersión de los datos?***
Conocer la dispersión es crucial en Inteligencia Artificial por varias razones:  
1. Confiabilidad del Promedio: Si la dispersión es muy alta, el promedio por sí solo no es un buen representante del conjunto de datos.
2. Detección de Anomalías (Outliers): Ayuda a identificar valores atípicos que se alejan demasiado de la distribución normal y que podrían sesgar o arruinar el entrenamiento de un modelo de Machine Learning.
3. Escalado de Datos: Muchos algoritmos de IA (como las Redes Neuronales o SVM) requieren que los datos estén normalizados o estandarizados (con media 0 y desviación estándar 1) para converger de manera eficiente. No se puede estandarizar sin conocer la dispersión.

## **3. Generación de Datos Aleatorios**

La generación de datos sintéticos es una técnica fundamental en la Inteligencia Artificial y la Ciencia de Datos. Nos permite simular escenarios reales, probar algoritmos y entrenar modelos cuando no disponemos de un dataset completo. NumPy facilita esta tarea mediante su módulo random.  

En este ejercicio, generaremos una muestra de datos que sigue una distribución normal utilizando la función np.random.normal(), la cual recibe tres parámetros clave:
* **loc:** Representa la media (el centro de la distribución).
* **scale:** Representa la desviación estándar (la dispersión o qué tan ancha es la campana).
* **size**: La cantidad de datos (muestras) a generar.

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Generación de los datos con distribución normal
# Se corrige la sintaxis añadiendo las comas y cerrando el paréntesis
datos = np.random.normal(
    loc=50,     # Media de 50
    scale=10,   # Desviación estándar de 10
    size=1000   # 1000 muestras
)

# 2. Graficar la distribución normal (Histograma)
plt.figure(figsize=(8, 5))
# bins=30 divide los datos en 30 barras, edgecolor mejora la visibilidad
plt.hist(datos, bins=30, color='skyblue', edgecolor='black', alpha=0.7)

# Configuraciones adicionales para que la gráfica se vea profesional
plt.title('Distribución Normal Sintética (Media=50, Desv=10)', fontsize=14)
plt.xlabel('Valores Generados', fontsize=12)
plt.ylabel('Frecuencia', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Mostrar la gráfica
plt.show()
```

## **Respuestas a las Preguntas1.**

### ***¿Qué significa distribución normal?***
También conocida como Distribución Gaussiana, es una distribución de probabilidad continua que tiene forma de campana simétrica (la famosa "Campana de Gauss"). En una distribución normal pura, la media, la mediana y la moda son exactamente iguales y se ubican en el centro de la curva. La mayor cantidad de observaciones se agrupan alrededor de este valor central, y la probabilidad de encontrar valores extremos disminuye gradualmente a medida que nos alejamos hacia las "colas" de la campana.  

### ***2. ¿Por qué muchos algoritmos asumen distribuciones aproximadamente normales?***
Existen varias razones fundamentales por las que la Inteligencia Artificial y la estadística confían en esta distribución:  
* ***El Teorema del Límite Central:*** Este teorema matemático establece que, cuando se suman variables aleatorias independientes (independientemente de su distribución original), su suma normalizada tiende a seguir una distribución normal si el tamaño de la muestra es lo suficientemente grande. Dado que muchos fenómenos del mundo real (como estaturas, errores de medición, salarios base) son el resultado de múltiples factores aleatorios, tienden a distribuirse de forma normal de manera natural.
* ***Simplicidad Matemática:*** Las funciones matemáticas asociadas a la distribución normal son fáciles de derivar y optimizar. Algoritmos como la Regresión Lineal, el Análisis de Componentes Principales (PCA) y ciertas redes neuronales asumen normalidad porque simplifica el cálculo de errores (como el Error Cuadrático Medio) y acelera la convergencia (el tiempo que tarda el modelo en aprender).

## **4. LIBRERÍA MATPLOTLIBMatplotlib**
es la biblioteca de visualización de datos más utilizada en el ecosistema de Python. Permite crear gráficos estáticos, animados e interactivos de alta calidad. Investigación de Concepto: Accuracy (Exactitud)
En el aprendizaje automático (Machine Learning) y Deep Learning, el accuracy es una métrica de evaluación que mide el rendimiento de un modelo de clasificación.
* ***Definición:*** Representa el porcentaje de predicciones correctas que realizó el modelo sobre el total de predicciones. Por ejemplo, un accuracy del 90% significa que el modelo acertó en 90 de cada 100 casos.
* ***Relación con las "Épocas" (Epochs):*** Una época representa un ciclo completo en el que el algoritmo de IA procesa todo el conjunto de datos de entrenamiento. Al graficar el accuracy frente a las épocas, podemos observar cómo el modelo va "aprendiendo" con el tiempo. Lo ideal es ver una curva ascendente que eventualmente se estabiliza, indicando que el modelo ha alcanzado su máxima capacidad de aprendizaje con los datos proporcionados.

```python
import matplotlib.pyplot as plt

# Datos de entrenamiento simulados
epocas = [1, 2, 3, 4, 5, 6, 7, 8]
accuracy = [60, 68, 75, 81, 85, 88, 90, 92]

# Configuración del tamaño de la figura
plt.figure(figsize=(8, 5))

# Creación del gráfico de líneas
# marker='o' añade puntos en cada iteración para visualizar mejor el dato exacto
plt.plot(epocas, accuracy, marker='o', color='green', linestyle='-', linewidth=2, markersize=8)

# Etiquetas y título
plt.title("Evolución del Accuracy por Época de Entrenamiento", fontsize=14, fontweight='bold')
plt.xlabel("Épocas (Ciclos de entrenamiento)", fontsize=12)
plt.ylabel("Accuracy (%)", fontsize=12)

# Añadir una cuadrícula para facilitar la lectura
plt.grid(True, linestyle='--', alpha=0.6)

# Renderizar y mostrar el gráfico
plt.show()
```

## ***5. Gráfico Histograma***

En esta sección, utilizaremos nuevamente la función ***np.random.normal()***  para generar un conjunto de datos sintéticos, pero esta vez nos enfocaremos netamente en su visualización utilizando un Histograma de Matplotlib ***(plt.hist()).***

A diferencia del gráfico de líneas que vimos en la sección anterior (ideal para ver tendencias en el tiempo o por épocas), el histograma agrupa los datos en rangos o intervalos (llamados bins) para mostrarnos cuántos datos caen dentro de cada rango.

```python
import numpy as np
import matplotlib.pyplot as plt

# Generación de datos simulados
# Media (loc) = 70, Desviación estándar (scale) = 15, Muestras (size) = 1000
datos_hist = np.random.normal(70, 15, 1000)

# Configuración del tamaño de la figura
plt.figure(figsize=(8, 5))

# Creación del histograma
# bins=30 define la cantidad de barras o intervalos. edgecolor dibuja el borde de las barras.
plt.hist(datos_hist, bins=30, color='coral', edgecolor='black', alpha=0.8)

# Configuraciones de título y etiquetas
plt.title("Distribución de Frecuencias (Histograma)", fontsize=14, fontweight='bold')
plt.xlabel("Rangos de Valores Generados", fontsize=12)
plt.ylabel("Frecuencia (Cantidad de datos)", fontsize=12)

# Añadir una cuadrícula horizontal para facilitar la lectura de las frecuencias
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Renderizar y mostrar el gráfico
plt.show()
```

## ***Respuestas a las Preguntas***

### ***1. ¿Qué representa un histograma?***
Un histograma es una representación gráfica de la distribución de frecuencias de un conjunto de datos numéricos continuos. En el eje X se representan los rangos o intervalos de los datos (las "clases"), y en el eje Y se representa la frecuencia, es decir, la cantidad de observaciones que caen dentro de cada uno de esos rangos. Visualmente se compone de barras adyacentes, lo que indica la continuidad de los datos.

### ***2. ¿Qué utilidad tiene para comprender un dataset?***
En el Análisis Exploratorio de Datos (EDA) para Inteligencia Artificial, el histograma es una de las herramientas más valiosas porque permite identificar a simple vista:

* La forma de la distribución: Nos dice si los datos siguen una distribución normal (campana simétrica), si están sesgados hacia la izquierda o la derecha, o si tienen múltiples picos (distribución bimodal).

* La tendencia central y la dispersión: Muestra dónde se concentra el grueso de la información y qué tan extendidas están las "colas" de los datos.

* Presencia de anomalías (Outliers): Permite detectar valores atípicos extremos que aparecen como barras muy pequeñas y aisladas lejos del bloque principal de datos. Estos valores atípicos a menudo deben ser tratados o eliminados antes de entrenar un modelo de Machine Learning.

## ***6. Gráfico de Barras***   

En los problemas de clasificación de Inteligencia Artificial (por ejemplo, identificar enfermedades en plantas), es común contar con diferentes "clases" o categorías. El gráfico de barras (plt.bar()) nos permite comparar visualmente la cantidad de registros o muestras que pertenecen a cada una de estas clases.

En este ejercicio, completaremos el código base para graficar la distribución de cuatro clases de plantas: 'Sano', 'Roya', 'Mosaico' y 'Mancha'.

```python
import matplotlib.pyplot as plt

# Definición de las categorías (clases) y sus respectivas cantidades
clases = ['Sano', 'Roya', 'Mosaico', 'Mancha'] #
cantidad = [1200, 850, 430, 320] # [cite: 123]

# Configuración del tamaño de la figura
plt.figure(figsize=(8, 5))

# Creación del gráfico de barras
# Se asigna un color personalizado y un ancho de barra estético
plt.bar(clases, cantidad, color=['#2ec4b6', '#e71d36', '#ff9f1c', '#011627'], edgecolor='black', alpha=0.85) # [cite: 124]

# Título y etiquetas de los ejes
plt.title("Distribución de Clases en el Dataset de Plantas", fontsize=14, fontweight='bold') # [cite: 125]
plt.xlabel("Estado de la Planta (Clase)", fontsize=12)
plt.ylabel("Cantidad de Muestras", fontsize=12)

# Añadir etiquetas de valor encima de cada barra para mayor precisión analítica
for i, val in enumerate(cantidad):
    plt.text(i, val + 20, str(val), ha='center', va='bottom', fontsize=10, fontweight='bold')

# Mostrar la cuadrícula solo en el eje Y
plt.grid(axis='y', linestyle='--', alpha=0.5)

# Renderizar y mostrar el gráfico
plt.show() # [cite: 126]
```

***Nota de análisis exploratorio:*** Al observar la gráfica, se evidencia un dataset desbalanceado. La clase 'Sano' cuenta con 1200 muestras, mientras que la clase 'Mancha' solo cuenta con 320. Al entrenar un modelo de IA con estos datos, este tenderá a ser muy bueno reconociendo plantas sanas, pero podría fallar clasificando la clase 'Mancha' debido a la escasez de ejemplos.

## ***7. LIBRERÍA SEABORN***
Seaborn es una biblioteca de visualización de datos de Python basada en Matplotlib. Está integrada estrechamente con las estructuras de datos de Pandas  y ofrece una interfaz de alto nivel para trazar gráficos estadísticos atractivos e informativos.  
* ***¿Qué son los mapas de calor (Heatmaps)?***
Es una representación gráfica de datos donde los valores individuales contenidos en una matriz se representan como colores. Las tonalidades cromáticas permiten identificar de forma instantánea intensidades, patrones o concentraciones de variables.  
* ***¿Por qué las matrices de correlación son importantes en Machine Learning?***
La correlación mide la relación lineal entre dos variables numéricas. Varía entre -1 y 1. Es crucial porque:
1. **Identifica variables clave:** Permite ver cuáles variables de entrada (características) están fuertemente relacionadas con nuestra variable objetivo (lo que queremos predecir)
2. **Detecta Multicolinealidad**: Si dos variables de entrada están demasiado correlacionadas entre sí (cercanas a 1 o -1), aportan información redundante. Eliminar una de ellas simplifica el modelo y mejora su rendimiento.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Creación del DataFrame corrigiendo la sintaxis del diccionario
datos = pd.DataFrame({
    'Edad': [20, 25, 30, 35, 40],
    'Horas': [2, 4, 6, 8, 10],
    'Nota': [60, 70, 80, 90, 100]
})

# 2. Calcular la matriz de correlación de Pearson
correlacion = datos.corr()

# Configuración del tamaño de la figura
plt.figure(figsize=(6, 4))

# 3. Dibujar el Heatmap
# annot=True muestra los valores numéricos dentro de cada celda
# cmap='coolwarm' asigna azul a correlaciones negativas y rojo a las positivas
# vmin y vmax aseguran que la escala vaya de -1 a 1
sns.heatmap(correlacion, annot=True, cmap='coolwarm', vmin=-1, vmax=1, linewidths=0.5)

# Título
plt.title("Mapa de Calor: Matriz de Correlación", fontsize=12, fontweight='bold')

# Mostrar la gráfica
plt.show()
```

***Interpretación del Gráfico:***
En este mapa de calor, observamos una correlación perfecta de 1.0 entre todas las variables (***Edad, Horas, Nota***). Esto se debe a que los datos de prueba crecen de manera perfectamente lineal y proporcional. En un escenario real de Inteligencia Artificial, valores tan perfectamente lineales son extremadamente raros, pero nos sirve para entender que a mayor intensidad del color rojo, más fuerte es la relación directa entre las variables.

## ***8. Gráfico Boxplot (Diagrama de Caja y Bigotes)***

El Boxplot es una representación gráfica basada en cuartiles que permite visualizar la distribución, la variabilidad y el centro de un conjunto de datos numéricos.

Este gráfico se compone de:

* La Caja: Representa el rango intercuartílico (IQR), que contiene el 50% central de los datos. La línea dentro de la caja es la mediana.

* Los Bigotes: Se extienden hasta los valores máximos y mínimos que no se consideran atípicos.

* Puntos aislados (si existen): Representan los valores atípicos (outliers), que caen fuera de los límites de los bigotes.

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Configuración del tamaño de la figura
plt.figure(figsize=(8, 4))

# Creación del Boxplot utilizando los datos del DataFrame anterior
# x=datos["Nota"] genera un boxplot horizontal para evaluar la variable 'Nota'
sns.boxplot(x=datos["Nota"], color='#ff9f1c', flierprops={"marker": "x", "markerfacecolor": "red", "markersize": 10}) #

# Título y etiquetas
plt.title("Diagrama de Caja (Boxplot) para la Variable 'Nota'", fontsize=14, fontweight='bold')
plt.xlabel("Notas", fontsize=12)

# Mostrar la gráfica
plt.show() # [cite: 143]
```

## **Respuestas a las Preguntas**
### 1. ***¿Qué son los valores atípicos (Outliers)?***
Son observaciones o puntos de datos que se encuentran a una distancia anormal de los demás valores en una muestra aleatoria. En otras palabras, son datos que "no siguen el patrón general" del resto del conjunto. En un Boxplot, matemáticamente se define como cualquier valor que se ubique por debajo de $Q_1 - 1.5 \times \text{IQR}$ o por encima de $Q_3 + 1.5 \times \text{IQR}$.  
### 2. ***¿Por qué pueden afectar el entrenamiento de un modelo?***
Los valores atípicos son una de las principales causas de un mal rendimiento en los modelos de Inteligencia Artificial por las siguientes razones:  
1. Sesgo en los algoritmos: Modelos basados en optimización de distancias o errores cuadráticos (como la Regresión Lineal, K-Means o Redes Neuronales) son altamente sensibles a los outliers. El modelo intentará ajustar sus parámetros para reducir el error de estos valores extremos, descuidando el patrón de la gran mayoría de los datos correctos.
2. Distorsión de métricas estadísticas: Afectan gravemente la media y la desviación estándar, haciendo que las técnicas de escalado de datos (como la estandarización) no funcionen de manera óptima.
3. Sobreajuste (Overfitting): El algoritmo puede memorizar el ruido o las anomalías en lugar de aprender las reglas generales del problema, reduciendo drásticamente su capacidad de generalizar con datos nuevos.

```python
import numpy as np
import pandas as pd

df = pd.read_csv(url)

# Definimos las columnas numéricas que vamos a evaluar
variables = ['Valor_Transaccion', 'Hora_Del_Dia', 'Numero_Transacciones_Diarias', 'Distancia_Geografica']

# Iniciamos el bucle para analizar cada columna una por una
for col in variables:
    # Guardamos los datos de la columna actual en una variable para no escribir df[col] tantas veces
    datos = df[col]


    conteo = np.size(datos)
    media = np.mean(datos)
    desv = np.std(datos)
    minimo = np.min(datos)
    p25 = np.percentile(datos, 25)
    p50 = np.percentile(datos, 50)
    p75 = np.percentile(datos, 75)
    maximo = np.max(datos)

    print(f"--- Estadísticas para: {col} ---")
    print(f"Count(Conteo): {conteo}") # Imprimimos el conteo total
    print(f"Mean (Media) : {media:.2f}")
    print(f"Std  (Desv)  : {desv:.2f}")
    print(f"Min  (Mínimo): {minimo:.2f}")
    print(f"25%  (Q1)    : {p25:.2f}")
    print(f"50%  (Mediana): {p50:.2f}")
    print(f"75%  (Q3)    : {p75:.2f}")
    print(f"Max  (Máximo): {maximo:.2f}")
    print("-" * 35) # Imprime una línea separadora para que se vea ordenado en la consola
```

#  **GENERACION DE GRAFICOS CON MATPLOBLIB**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Cargar el dataset
df = pd.read_csv(url)

# Preparación previa: Para el Heatmap, necesitamos que 'Fraude' sea una variable numérica (1 y 0)
df['Fraude_Num'] = df['Fraude'].map({'SI': 1, 'NO': 0})

# 1. Histogramas (Distribución de una variable numérica)
plt.figure(figsize=(8, 5))
plt.hist(df['Valor_Transaccion'], bins=30, color='skyblue', edgecolor='black')
plt.title('Histograma: Distribución del Valor de las Transacciones')
plt.xlabel('Valor de la Transacción')
plt.ylabel('Frecuencia')
plt.show()
```

#**INTERPRETACIÓN RESULTADOS**
El histograma agrupa los datos para mostrarnos dónde se concentra la mayor cantidad de información. Nos muestra cuántas transacciones se hicieron por cada rango de dinero.

Existe una distribución uniforme, al observar las barras, en general se mantiene entre 10 y 25 transacciones sin importar el monto. Hay tantas transacciones de 500,000 como transacciones de 4 millones.

```python
# 2. Gráficos de barras (Conteo de transacciones fraudulentas vs normales)
plt.figure(figsize=(6, 4))
df['Fraude'].value_counts().plot(kind='bar', color=['green', 'red'])
plt.title('Gráfico de Barras: Proporción de Fraudes')
plt.xlabel('¿Es Fraude?')
plt.ylabel('Cantidad de Transacciones')
plt.xticks(rotation=0)
plt.show()
```

# **INTERPRETACIÓN RESULTADOS**
Existe una distribución balanceada de los datos, el gráfico muestra que la barra verde ("SI" es fraude) y la barra rojo ("NO" es fraude) alcanzan los 250 registros. Esto significa que hay una proporción perfecta del 50% para transacciones fraudulentas y 50% para legítimas.

```python
# 3. Gráficos de dispersión (Relación entre dos variables)
plt.figure(figsize=(8, 6))
# Asignamos color rojo a los fraudes y azul a las transacciones normales
colores = np.where(df['Fraude'] == 'SI', 'red', 'blue')
plt.scatter(df['Valor_Transaccion'], df['Distancia_Geografica'], c=colores, alpha=0.5)
plt.title('Dispersión: Valor vs Distancia Geográfica (Rojo = Fraude)')
plt.xlabel('Valor de la Transacción')
plt.ylabel('Distancia Geográfica (km)')
plt.show()
```

# **INTERPRETACIÓN RESULTADOS**

El gráfico nos muestra que las transacciones normales suelen involucrar poco dinero y hacerse cerca de casa, mientras que los fraudes tienden a ser por montos altos y a grandes distancias.

#  **GENERACION DE ANÁLISIS VISUALES CON SEABORN**

```python
# 1. Gráficos de correlaciones (Heatmap)
plt.figure(figsize=(8, 6))
# Seleccionamos solo las variables numéricas
columnas_corr = ['Valor_Transaccion', 'Hora_Del_Dia', 'Numero_Transacciones_Diarias', 'Distancia_Geografica', 'Fraude_Num']
sns.heatmap(df[columnas_corr].corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Heatmap: Matriz de Correlaciones')
plt.show()
```

# **INTERPRETACIÓN RESULTADOS**


El mapa de calor traduce las relaciones visuales en números estadísticos (Coeficiente de Correlación de Pearson), yendo de -1 a 1. fila/columna (Fraude_Num):

Fuerte Predictor - Número de Transacciones Diarias (0.44): Es la variable que tiene la correlación positiva más fuerte con el fraude. Matemáticamente confirma que, a mayor cantidad de transacciones en un día, mayor es la probabilidad de fraude.

Predictor Moderado - Valor de Transacción (0.32): El monto del dinero también tiene una correlación positiva considerable. Valores más altos están asociados con casos fraudulentos.

Predictor Leve - Distancia Geográfica (0.25): Existe una relación positiva, indicando que transacciones más lejanas tienen más riesgo, pero no es tan determinante por sí sola como las dos anteriores.

Variable Irrelevante - Hora del Día (0.05): Una correlación tan cercana a 0 indica que la hora en la que se hace la transacción no tiene un impacto significativo para predecir si es fraude o no en este conjunto de datos específico.

Nos revela que la pista principal y más delatadora es el número de operaciones al día: hacer muchas transacciones seguidas es la mayor señal de alerta. La segunda pista más clara es la cantidad de dinero (montos muy altos son sospechosos), y la tercera es la distancia (compras muy lejanas aumentan un poco el riesgo). Por el contrario, el gráfico nos demuestra que la hora del pago no sirve como pista; en estos datos, saber a qué hora ocurrió la compra no ayuda en absolutamente nada para adivinar si es un fraude o una operación normal.
