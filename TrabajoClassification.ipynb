## **¿Qué es machine learning?**

El machine learning es el subconjunto de la inteligencia artificial (IA) centrado en algoritmos que pueden "aprender" los patrones de los datos de entrenamiento y, posteriormente, hacer inferencias precisas sobre nuevos datos. Esta capacidad de reconocimiento de patrones permite a los modelos de machine learning tomar decisiones o realizar predicciones sin instrucciones explícitas y codificadas de forma rígida.


## **¿Qué son los modelos de clasificación?**


Los modelos de clasificación son algoritmos que asignan una etiqueta o una categoría a un dato de entrada. Para ello, van comparando los patrones que hay dentro de un conjunto de datos etiquetados y así aprenden las características que tiene cada uno. Se trata de un aprendizaje supervisado previo, que les sirve para hacer futuras clasificaciones de los datos nuevos que reciben con una probabilidad o una decisión directa.

Estos modelos se usan en el aprendizaje automático para resolver tareas como clasificar textos, identificar imágenes, anticipar riesgos o comparar el comportamiento de los clientes.

## **¿Para qué se utilizan los modelos de clasificación?**

- Banca y finanzas: detección de fraude, evaluación de riesgo crediticio.

- Sanidad: diagnóstico de enfermedades a partir de imágenes o historiales.

- Marketing: clasificación de clientes según comportamiento o perfil.

- Industria: clasificación de productos defectuosos o alertas críticas.

## **Entrenamiento y evaluación de modelos**

En esta sección se implementan los algoritmos de clasificación seleccionados para el proyecto.

Con el propósito de mantener un código organizado y reutilizable, se desarrolla una función general denominada **evaluar_modelo()**, encargada de entrenar cada algoritmo, generar predicciones y calcular automáticamente las principales métricas de desempeño.

La función permite evaluar todos los modelos bajo los mismos criterios, facilitando posteriormente la comparación de resultados.

```python
# ==========================================================
# FUNCIÓN GENERAL DE EVALUACIÓN
# ==========================================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    ConfusionMatrixDisplay,
    roc_curve,
    roc_auc_score
)

resultados = []

def evaluar_modelo(modelo,
                   nombre,
                   X_train,
                   y_train,
                   X_test,
                   y_test):

    print("="*70)
    print(f"MODELO: {nombre}")
    print("="*70)

    # -----------------------------
    # Entrenamiento
    # -----------------------------
    modelo.fit(X_train, y_train)

    # -----------------------------
    # Predicciones
    # -----------------------------
    y_pred = modelo.predict(X_test)

    # -----------------------------
    # Probabilidades para ROC
    # -----------------------------
    if hasattr(modelo, "predict_proba"):
        y_prob = modelo.predict_proba(X_test)[:, 1]
    else:
        y_prob = modelo.decision_function(X_test)

    # -----------------------------
    # Métricas
    # -----------------------------
    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred)
    recall = recall_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)
    auc = roc_auc_score(y_test, y_prob)

    # -----------------------------
    # Guardar resultados
    # -----------------------------
    resultados.append({
        "Modelo": nombre,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1": f1,
        "AUC": auc
    })

    # -----------------------------
    # Mostrar métricas
    # -----------------------------
    print(f"Accuracy : {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall   : {recall:.4f}")
    print(f"F1 Score : {f1:.4f}")
    print(f"AUC      : {auc:.4f}")

    # -----------------------------
    # Classification Report
    # -----------------------------
    print("\nCLASSIFICATION REPORT")
    print(classification_report(y_test, y_pred))

    # -----------------------------
    # Matriz de Confusión
    # -----------------------------
    fig, ax = plt.subplots(figsize=(6,5))

    ConfusionMatrixDisplay.from_predictions(
        y_test,
        y_pred,
        cmap="Blues",
        ax=ax
    )

    plt.title(f"Matriz de Confusión\n{nombre}")
    plt.show()

    # -----------------------------
    # Valores de la Matriz de Confusión
    # -----------------------------
    cm = confusion_matrix(y_test, y_pred)

    tn, fp, fn, tp = cm.ravel()

    print("\n" + "="*70)
    print("VALORES DE LA MATRIZ DE CONFUSIÓN")
    print("="*70)

    print(f"Verdaderos Positivos (TP): {tp}")
    print(f"Verdaderos Negativos (TN): {tn}")
    print(f"Falsos Positivos (FP): {fp}")
    print(f"Falsos Negativos (FN): {fn}")

    print("\nINTERPRETACIÓN")

    print(f"""
TP (Verdaderos Positivos): {tp}
→ El modelo clasificó correctamente a {tp} estudiantes que realmente aprobaron.

TN (Verdaderos Negativos): {tn}
→ El modelo clasificó correctamente a {tn} estudiantes que realmente reprobaron.

FP (Falsos Positivos): {fp}
→ El modelo predijo que {fp} estudiantes aprobarían cuando en realidad reprobaron.
Este corresponde a un Error Tipo I.

FN (Falsos Negativos): {fn}
→ El modelo predijo que {fn} estudiantes reprobarían cuando en realidad aprobaron.
Este corresponde a un Error Tipo II.
""")

    # -----------------------------
    # Curva ROC
    # -----------------------------
    fpr, tpr, _ = roc_curve(y_test, y_prob)

    plt.figure(figsize=(7,6))

    plt.plot(
        fpr,
        tpr,
        linewidth=3,
        label=f"AUC = {auc:.3f}"
    )

    plt.plot(
        [0,1],
        [0,1],
        linestyle="--",
        color="red"
    )

    plt.xlabel("False Positive Rate")
    plt.ylabel("True Positive Rate")
    plt.title(f"Curva ROC - {nombre}")
    plt.legend()
    plt.grid(True)
    plt.show()

    # -----------------------------
    # Retornar resultados
    # -----------------------------
    return {
        "modelo": modelo,
        "accuracy": accuracy,
        "precision": precision,
        "recall": recall,
        "f1": f1,
        "auc": auc
    }
```

## **Análisis de Overfitting y Underfitting**

Uno de los principales desafíos en Machine Learning consiste en lograr un equilibrio entre la capacidad del modelo para aprender los patrones presentes en los datos de entrenamiento y su capacidad para generalizar sobre datos no vistos.

En esta sección se compara el desempeño del mejor modelo tanto en el conjunto de entrenamiento como en el conjunto de prueba, con el propósito de identificar posibles evidencias de sobreajuste (Overfitting) o subajuste (Underfitting).

```python
# ==========================================================
# Evaluación del modelo en entrenamiento y prueba
# ==========================================================

from sklearn.metrics import accuracy_score

# Predicciones en entrenamiento
y_train_pred = mejor_dt.predict(X_train)

# Predicciones en prueba
y_test_pred = mejor_dt.predict(X_test)

# Accuracy
train_accuracy = accuracy_score(y_train, y_train_pred)
test_accuracy = accuracy_score(y_test, y_test_pred)

print("="*60)
print("RENDIMIENTO DEL MODELO")
print("="*60)

print(f"Accuracy entrenamiento : {train_accuracy:.4f}")
print(f"Accuracy prueba        : {test_accuracy:.4f}")
```

```python
# ==========================================================
# Comparación de Accuracy
# ==========================================================

comparacion = pd.DataFrame({

    "Conjunto":["Entrenamiento","Prueba"],

    "Accuracy":[train_accuracy,test_accuracy]

})

plt.figure(figsize=(7,5))

sns.barplot(

    data=comparacion,

    x="Conjunto",

    y="Accuracy",

    palette="viridis"

)

plt.ylim(0.7,1)

plt.title("Comparación del Accuracy")

plt.show()
```

```python
# ==========================================================
# Diferencia entre entrenamiento y prueba
# ==========================================================

diferencia = train_accuracy - test_accuracy

print(f"Diferencia: {diferencia:.4f}")
```

```python
# ==========================================================
# Diagnóstico del modelo
# ==========================================================

if diferencia < 0.03:

    print("Excelente capacidad de generalización.")
    print("No se observan evidencias importantes de overfitting.")

elif diferencia < 0.07:

    print("Existe un ligero overfitting.")
    print("El modelo aún puede considerarse adecuado.")

else:

    print("El modelo presenta overfitting.")
    print("Sería recomendable reducir la complejidad del árbol.")
```

```python
# ==========================================================
# Accuracy vs profundidad
# ==========================================================

profundidades = range(1,21)

train_scores = []

test_scores = []

for d in profundidades:

    modelo = DecisionTreeClassifier(

        max_depth=d,

        random_state=42

    )

    modelo.fit(X_train,y_train)

    train_scores.append(

        modelo.score(X_train,y_train)

    )

    test_scores.append(

        modelo.score(X_test,y_test)

    )

plt.figure(figsize=(10,6))

plt.plot(

    profundidades,

    train_scores,

    marker="o",

    label="Entrenamiento"

)

plt.plot(

    profundidades,

    test_scores,

    marker="s",

    label="Prueba"

)

plt.xlabel("Profundidad máxima")

plt.ylabel("Accuracy")

plt.title("Overfitting y Underfitting")

plt.legend()

plt.grid()

plt.show()
```

## Interpretación de resultados

La gráfica muestra cómo cambia el desempeño del modelo a medida que aumenta la profundidad máxima del árbol de decisión.

Cuando la profundidad es muy pequeña, el modelo no logra capturar adecuadamente los patrones presentes en los datos, lo que se traduce en un bajo desempeño tanto en entrenamiento como en prueba. Esta situación corresponde a un caso de **Underfitting**.

A medida que aumenta la profundidad del árbol, el desempeño mejora tanto en entrenamiento como en prueba hasta alcanzar un punto óptimo.

Si la profundidad continúa aumentando, el modelo comienza a ajustarse excesivamente a los datos de entrenamiento, obteniendo una precisión muy alta en este conjunto pero disminuyendo su capacidad de generalización sobre datos nuevos. Este comportamiento corresponde al fenómeno conocido como **Overfitting**.

El modelo seleccionado mediante GridSearchCV presenta un equilibrio adecuado entre ambas situaciones, logrando una buena capacidad predictiva sin evidencias significativas de sobreajuste.
